```python
# Leitura dos dados da PA
primeiro_termo = int(input("Digite o primeiro termo da PA: "))
razao = int(input("Digite a razão da PA: "))

termo = primeiro_termo
total_mostrados = 0
mais_termos = 10

# Mostra os termos enquanto o usuário quiser
while mais_termos != 0:
    contador = 0
    while contador < mais_termos:
        print(termo)
        termo += razao
        contador += 1
        total_mostrados += 1

    print(f"Já foram mostrados {total_mostrados} termos.")
    mais_termos = int(input("Quantos termos você quer mostrar a mais? (0 para encerrar): "))

print("Programa encerrado.")
```
